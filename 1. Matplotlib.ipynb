Matplotlib Basics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,20,10)
y = x ** 2
```

```python
x
```

```python
y
```

```python
plt.plot(x,y)
plt.title('My first Graph')
plt.xlabel("Chacha")
plt.ylabel("Chahchi")
```

Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,x)
plt.subplot(2,2,4)
plt.plot(x,x**3)
```

Object Oriented Way

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

    figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch(pixel-per-pixel) 

```python
fig = plt.figure(figsize = (12,12), dpi = 800)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```

Types Of Plots

Scatter plot

```python
plt.scatter(x,y)
```

Histogram

```python
plt.hist(x,y)
```

Box plot

```python
plt.boxplot(x)
```

Saving Plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('Basic plot.png')
```
